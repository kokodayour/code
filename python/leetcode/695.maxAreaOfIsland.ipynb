```python
def maxAreaOfIsland(grid):
    m, n = len(grid), len(grid[0])
    ans = 0
    def dfs(i, j):
        nonlocal area
        if (i<0) or (i>=m) or (j<0) or (j>=n) or (grid[i][j] != 1):
            return
        grid[i][j] = 2 # avoid repeated traversal
        area += 1
        dfs(i, j-1) # go left
        dfs(i, j+1) # go right
        dfs(i+1, j) # go up
        dfs(i-1, j) # go down
    for i, row in enumerate(grid):
        for j, col in enumerate(row):
            if col == 1: # find new island
                area = 0
                dfs(i, j)
                ans = max(ans, area)
    return ans
```
