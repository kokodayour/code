```python
def minimumCost(cost):
    ans = 0
    cost.sort(reverse=True)
    for x, y, z in zip(cost[::3], cost[1::3], cost[2::3]):
        ans += x
        ans += y
    left = len(cost)%3
    if left == 2:
        return ans+cost[-1]+cost[-2]
    elif left == 1:
        return ans+cost[-1]
    else:
        return ans
```

```python
def minimumCost(cost):
    ans = 0
    n = len(cost)
    cost.sort()
    for i in range(n):
        if i%3 == 2:
            continue
        # 第i个数和倒数第i个数索引之和是n-1
        ans += cost[n-1-i]
    return ans
```
