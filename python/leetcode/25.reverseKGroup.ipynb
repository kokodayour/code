```python
from node import *
# 错误写法 当剩余的元素个数不足k的时候也会翻转
def reverseKGroup(head, k: int):
    dummy = pre = ListNode(next=head)
    while pre.next:
        cur = pre.next
        for _ in range(k-1):
            if not cur.next:
                break
            next = cur.next
            cur.next = next.next
            next.next = pre.next
            pre.next = next
        pre = cur
    return dummy.next
```

```python
def reverseKGroup(head, k: int):
    dummy = pre = ListNode(next=head)
    while pre.next:
        cur = pre.next
        for _ in range(k-1):
            nxt = cur.next
            cur.next = nxt.next
            nxt.next = pre.next
            pre.next = nxt
        temp = pre = cur
        # determine whether there are remaining K elements
        for _ in range(k):
            temp = temp.next
            if not temp:
                return dummy.next  
    return dummy.next
```
