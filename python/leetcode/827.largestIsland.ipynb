```python
def largestIsland(grid):
    n = len(grid)
    def dfs(i, j):
        size = 1
        grid[i][j] = len(area) + 2 # record (i, j) belong to which island
        for x, y in (i-1,j), (i+1,j), (i,j-1), (i,j+1):
            if (0<=x<n) and (0<=y<n) and (grid[x][y]==1):
                size += dfs(x, y)
        return size
    
    # DFS each island to count the area of each, append them to area array
    area = []
    for i, row in enumerate(grid):
        for j, x in enumerate(row):
            if x == 1:
                area.append(dfs(i, j))

    if not area:
        return 1
    
    ans = 0
    for i, row in enumerate(grid):
        for j, x in enumerate(row):
            if x: continue
            s = set()
            for x, y in (i-1, j), (i+1, j), (i, j-1), (i, j+1):
                if 0 <= x < n and 0 <= y < n and grid[x][y]:
                    s.add(grid[x][y])
            ans = max(ans, sum(area[idx-2] for idx in s) + 1)
    
    # if ans is still 0, all grids are 1, return n2
    return ans if ans else n*n
```
