```python
def rotate(nums: list[int], k: int) -> None:
    """
    Do not return anything, modify nums in-place instead.
    """
    n = len(nums)
    k = n
    temp = nums.copy()
    for i in range(k,n):
        nums[i] = temp[i-k ]
    for i in range(k):
        nums[i] = temp[-k+i]
    return nums
```

```python
def rotate(nums: list[int], k: int) -> None:
    """
    Do not return anything, modify nums in-place instead.
    """
    cut = len(nums)-k%len(nums)
    nums[:] = nums[cut:]+nums[:cut]
    return nums
```

```python
nums = [1,2,3,4,5,6,7]
k = 3
rotate(nums,k)
```
