```python
def flatten(root):
    """
    Do not return anything, modify root in-place instead
    """
    result = []
    def f(node):
        """
        preorder
        """
        if node:
            result.append(node)
            f(node.left)
            f(node.right)
    f(root)
    for x1, x2 in zip(result, result[1:]):
        x1.left = None
        x1.right = x2
```

```python
head = None
def flatten(root):
    """
    head insertion
    right => left => root
    """
    if root is None:
        return
    flatten(root.right)
    flatten(root.left)
    root.left = None
    root.right = head
    head = root
```

```python
def flatten(root):
    """
    divide and conquer
    return the tail element of linkedlist
    """
    if root is None:
        return None
    left_tail = flatten(root.left)
    right_tail = flatten(root.right)
    if left_tail:
        left_tail.right = root.right
        root.right = root.left
        root.left = None
    return right_tail or left_tail or root
```
