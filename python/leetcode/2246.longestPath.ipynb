```python
def longestPath(parent, s):
    n = len(parent)
    g = [[] for _ in range(n)]
    # build tree
    for i in range(1, n):
        g[parent[i]].append(i)
    
    ans = 0
    # return the maximum number of paths from this node
    def dfs(x: int) -> int:
        nonlocal ans
        max_len = 0
        for y in g[x]:
            # when y is equal to x, max_len will not update later and len will be meaningless  
            # It must be outside the if statement, because ans maybe in the subtree.
            # If it is inside the if statement, dfs will call only once when s[y] == s[x] and you will get 0+
            len = dfs(y) + 1
            if s[y] != s[x]:
                # max_len and len are the maximum and second maximum respectively
                ans = max(ans, max_len+len)
                # update max_len
                max_len = max(max_len, len)
        return max_len
    dfs(0)
    return ans + 1
```

```python
parent = [-1,0,0,1,1,2]
s = "abacbe"
# parent = [-1,0,0,0]
# s = "aabc"
longestPath(parent, s)
```
