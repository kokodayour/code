```python
def removeNthFromEnd(head, n):
    # number of all nodes
    num = 0
    cur = head
    while cur:
        num += 1
        cur = cur.next
    dummy = pre = ListNode(next=head)
    for _ in range(num-n):
        pre = pre.next
    pre.next = pre.next.next
    # cannot return head directly, for head maybe deleted
    return dummy.next
```

```python
# speed_pointer
def removeNthFromEnd(head, n):
    # 
    slow = fast = dummy = ListNode(next=head)
    for _ in range(n):
        fast = fast.next
    while fast.next:
        slow = slow.next
        fast = fast.next
    slow.next = slow.next.next
    return dummy.next
```
