```python
def partition(s:str) -> list[list[str]]:
    n = len(s)
    ans = []
    subs = []

    def dfs(i):
        """
        i is the index of comma
        """
        if i == n:
            ans.append(subs.copy())
            return
        
        # use comma to split str, select 0 to n-1 commas which can get all substring
        for comma in range(i, n):
            substring = s[i: comma+1]
            # judge whether substring is 回文串
            if substring == substring[::-1]:
                subs.append(substring)
                dfs(comma+1)
                subs.pop()

    dfs(0)
    return ans
```

```python
def partition(s: str) -> list[list[str]]:
    n = len(s)
    ans = []
    subs = []

    def dfs(comma, start):
        """
        Args:
            comma: comma after i
            start: the index of first elem of 回文串
        """
        if comma == n:
            ans.append(subs.copy())
            return
        # not select
        # n-1 is the last elem and we must select it, otherwise the last part will not append to subs
        if comma < n-1:
            dfs(comma+1, start)
        # select
        substring = s[start: comma+1]
        if substring == substring[::-1]:
            subs.append(substring)
            dfs(comma+1, comma+1)
            subs.pop()
        
    dfs(0, 0)
    return ans
```
