```python
def moveZeroes(nums: list[int]) -> None:
    for i, x in enumerate(nums):
        if x == 0:
            nums.pop(i)
            nums.append(0)
    return nums
```

```python
nums = [0,1,0,3,12]
nums = [0,0,1]
moveZeroes(nums)
```
