```python
def lowestCommonAncestor(root, p, q):
    """
    return the lowestCommonAncestor
    """
    if (root is None) or (root is p) or (root is q):
        return root
    left = lowestCommonAncestor(root.left, p, q)
    right = lowestCommonAncestor(root.right, p, q)
    if left and right:
        return root
    if left:
        return left
    return right
```
