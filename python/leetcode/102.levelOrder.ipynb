```python
def levelOrder(root):
    if not root:
        return []
    ans = []
    queue = [root] # 当层节点
    while queue:
        temp = [] # 节点值
        nodes = queue
        queue = [] # 清空当层节点 存放下层节点
        for node in nodes:
            temp.append(node.val)
            if node.left:
                queue.append(node.left)
            if node.right:
                queue.append(node.right)
        ans.append(temp)
    return ans
```
