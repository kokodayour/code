```python
def islandPerimeter(grid):
    m, n = len(grid), len(grid[0])
    ans = 0
    for i, row in enumerate(grid):
        for j, x in enumerate(row):
            if x == 1:
                # if one direction is sea or out of boundary, perimeter in that direction add 1
                if i-1 == -1 or grid[i-1][j] == 0: # go up
                    ans += 1
                if i+1 == m or grid[i+1][j] == 0: # go down
                    ans += 1
                if j-1 == -1 or grid[i][j-1] == 0: # go left
                    ans += 1
                if j+1 == n or grid[i][j+1] == 0: # go right
                    ans += 1
    return ans
```
