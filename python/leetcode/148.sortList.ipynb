```python
from node import *

def sortList(head):
    if not head:
        return None
    a = []
    cur = head
    while cur:
        a.append(cur)
        cur = cur.next
    a.sort(key=lambda x: x.val)
    head = cur = a[0]
    for x in a[1:]:
        cur.next = x
        cur = cur.next
    cur.next = None
    return head
```
