```python
# refer to question 46.permute
def solveNQueens(n):
    ans = []
    # each elem is the index of chessman in each row
    col = [0]*n

    # judge whether the cur chessman conflicts with the previous chessmen
    def valid(r ,c):
        for R in range(r):
            C = col[R]
            # judge whether the cur chessman is on the same digonal line with the pre chessmen
            if r+c == R+C or r-c == R-C:
                return False
        return True

    # r is the next row and s is the remaining cols
    def dfs(r, s):
        # have iterated all rows
        if r == n:
            # create chessboard
            ans.append(['.'*c + 'Q' + '.'*(n-1-c) for c in col])
            return
        for c in s:
            # another way to replace valid func
            # if all(r+c != R+col[R] and r-c != R-col[R] for R in range(r)):
            if valid(r, c):
                col[r] = c
                dfs(r+1, s-{c})
    dfs(0, set(range(n)))
    return ans
```

```python
def solveNQueens(n):
    ans = []
    col = [0]*n
    on_path = [False]*n
    m = 2*n-1
    # for there are 2N-1 r+c and r-c
    diag1 = [False]*m
    diag2 = [False]*m

    def dfs(r):
        if r == n:
            ans.append(['.'*c + 'Q' + 'c'*(n-1-c) for c in col])
            return
        for c in range(n):
            if not on_path[c] and not diag1[r+c] and not diag2[r-c]:
                col[r] = c
                on_path[c] = diag1[r+c] = diag2[r-c] = True
                dfs(r+1)
                on_path[c] = diag1[r+c] = diag2[r-c] = False
    dfs(0)
    return ans
```
