```python
def permute(nums):
    n = len(nums)
    ans = []
    path = [0]*n
    def dfs(i, s):
        if i == n:
            ans.append(path.copy())
            return
        for x in s:
            path[i] = x
            dfs(i+1, s-{x})
    dfs(0, set(nums))
    return ans
```

```python
def permute(nums):
    n = len(nums)
    ans = []
    path = [0]*n 
    on_path = [False] * n
    def dfs(i):
        if i == n:
            ans.append(path.copy())
            return 
        for j in range(n):
            if not on_path[j]:
                path[i] = nums[j]
                on_path[j] = True
                dfs(i+1)
                on_path[j] = False
    dfs(0)
    return ans
```

```python
def permute(nums: list[int]) -> list[list[int]]:
    ans = []
    n = len(nums)
    path = [0]*n
    lookup = {}
    for i in range(n):
        lookup[nums[i]] = False
    def dfs(i):
        if i == n:
            ans.append(path.copy())
            return
        for x in nums:
            if lookup[x] == True: continue
            path[i] = x
            lookup[x] = True
            dfs(i+1)
            lookup[x] = False
    dfs(0)
    return ans
```

```python
permute([1,2,3])
```
