```python
def numIslands(grid):
    m, n = len(grid), len(grid[0])
    def dfs(i, j):
        if (i < 0) or (i >= m) or (j < 0) or (j >= n) or grid[i][j] != '1':
            return
        grid[i][j] = '2' # aovid repeated traversal
        dfs(i, j-1) # go left
        dfs(i, j+1) # go right
        dfs(i+1, j) # go up
        dfs(i-1, j) # go down
    ans = 0
    for i, row in enumerate(grid):
        for j, col in enumerate(row):
            if col == '1': # find a new island
                dfs(i, j)
                ans += 1
    return ans
```
