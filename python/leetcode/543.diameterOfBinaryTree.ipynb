```python
# 获取树的深度
def get_depth(root):
    return max(get_depth(root.left), get_depth(root.right))+1 if root else 0
# 获取经过根节点的直径
def get_diameter(root):
    left = get_depth(root.left)
    right = get_depth(root.right)
    return left + right
def diameterOfBinaryTree(root):
    ans = 0
    def func(root):
        nonlocal ans
        if root:
            ans = max(ans, get_diameter(root))
            func(root.left)
            func(root.right)
    return ans
```
