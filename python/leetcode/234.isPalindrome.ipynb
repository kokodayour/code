```python
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next

def isPalindrome(head: Optional[ListNode]) -> bool:
    a = []
    while head:
        a.append(head.val)
        head = head.next
    b = list(reversed(a))
    n = len(a)
    n = n//2 if n%2==0 else (n-1)//2
    for x1, x2 in zip(a[:n],b[:n]):
        if x1 == x2:
            continue
        else:
            return False
    return True
```
