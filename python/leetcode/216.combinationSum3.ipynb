```python
# the sum of k numbers is n
def combinationSum3(self, k: int, n: int):
    ans = []
    path = [0] * k

    def dfs(start, i, current_sum):
        if i == k:
            if current_sum == n:
                ans.append(path.copy())
            return
        for x in range(start, 10 - k + i + 1):
            # must use temporary var to avoid modify current_sum otherwise it will be larger after every iteration
            new_sum = current_sum + x 
            if new_sum > n:
                break
            path[i] = x
            dfs(x + 1, i + 1, new_sum)  

    dfs(1, 0, 0)
    return ans
```

```python
combinationSum3(3, 9)
```
