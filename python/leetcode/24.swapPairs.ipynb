```python
from node import *
def swapPairs(head):

    node0 = dummy = ListNode(next=head)
    while node0.next and node0.next.next:
        node1 = node0.next
        node2 = node0.next.next
        node3 = node0.next.next.next
        node0.next = node2
        node1.next = node3
        node0.next.next = node1
        node0 = node0.next.next
    return dummy.next
```
