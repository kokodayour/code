```python
def rightSideView(root: Optional[TreeNode]) -> List[int]:
    ans = []
    # 怎么判断该节点是否应该被记录
    # 当该节点的深度等于答案长度时
    # 递归函数将深度不断地传下去
    # 先遍历右 在遍历左
    def f(node, depth):
        if node is None:
            return 
        if depth == len(ans):
            ans.append(node.val)
        f(node.right, depth+1)
        f(node.left, depth+1)
    f(root, 0)
    return ans
```

```python
# 深搜
def rightSideView(root):
    lookup = {} # {深度: node.val}
    max_depth = -1
    st = [(root, 0)]
    while st:
        # 先看右边
        node, depth = st.pop()
        if node:
            max_depth = max(max_depth, depth)

            # 如果不存在对应深度的节点才插入
            # 也就是说只保留每层看到的第一个节点
            lookup.setdefault(depth, node.val)
            st.append((node.left, depth+1))
            st.append((node.right, depth+1))
    return lookup.values()
```

```python
# 层序遍历 每一层最后一个被访问的节点就是右视图
def rightSideView(root):
    if not root:
        return []

    ans = []
    q = deque([root])
    while q:
        level_size = len(q)
        for i in range(level_size):
            node = q.popleft()
            if node.left:
                q.append(node.left)
            if node.right:
                q.append(node.right)
            # 只在遍历到当前层最后一个节点时记录
            if i == level_size - 1:
                ans.append(node.val)
    return ans
```
