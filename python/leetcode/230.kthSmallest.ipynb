```python
def kthSmallest(root, k) -> int:
    # if find kthSmallest, pass it up, else pass None up
    # k-1 while each iteration
    def f(node):
        nonlocal k
        if node is None:
            return None
        else:
            result = f(node.left)
            if result is not None:
                return result
            k -= 1
            if k == 0:
                return node.val
            f(node.right)
    return f(root)
```
