```python
def subsets(nums: list[int]) -> list[list[int]]:
    n = len(nums)
    ans = []
    path = []

    def dfs(start):
        for i in range(start, n):
            path.append(nums[i])
            ans.append(path.copy())
            dfs(i+1)
            path.pop()

    dfs(0)
    return ans+[[]]
```

```python
# choose or not for every input number in nums
def subsets(nums):
    n = len(nums)
    ans = []
    path = []

    def dfs(i):
        if i == n:
            ans.append(path.copy())
            return
        # not choose
        dfs(i+1)
        # choose
        path.append(nums[i])
        dfs(i+1)
        path.pop()

    dfs(0)
    return ans
```

```python
subsets(nums = [1,2,3])
```
