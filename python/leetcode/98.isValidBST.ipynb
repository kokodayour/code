```python
# levelOrder
def isValidBST(root):
    p_inf = float('inf')
    n_inf =float('-inf')
    # nodes to be traversed and its left and right number
    queue = [(n_inf, root, p_inf)]
    while queue:
        left, node, right = queue.pop(0)
        if node.left:
            # can it insert the left side
            if left < node.left.val < node.val:
                queue.append((left, node.left, node.val))
            else:
                return False
        if node.right:
            # can it insert the right side
            if node.val < node.right.val < right:
                queue.append((node.val, node.right, right))
            else:
                return False
    return True
```
