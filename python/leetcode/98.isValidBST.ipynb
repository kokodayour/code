```python
# levelOrder
def isValidBST(root) -> bool:
    p_inf = float('inf')
    n_inf =float('-inf')
    queue = deque([(n_inf, root, p_inf)])
    while queue:
        left, node, right = queue.popleft()
        # determine the next one before inserting
        if node.left:
            # can it insert the left side
            if left < node.left.val < node.val:
                queue.append((left, node.left, node.val))
            else:
                return False
        if node.right:
            # can it insert the right side
            if node.val < node.right.val < right:
                queue.append((node.val, node.right, right))
            else:
                return False
    return True
```

```python
# levelOrder(SP)
from collections import deque
def isValidBST(root):
    if not root:
        return True
    
    p_inf = float('inf')
    n_inf = float('-inf')
    queue = deque([(n_inf, root, p_inf)])
    
    while queue:
        left, node, right = queue.popleft()  # O(1)
        # determine this one and insert the next
        if not (left < node.val < right):
            return False            
        if node.left:
            queue.append((left, node.left, node.val))
        if node.right:
            queue.append((node.val, node.right, right))
    return True
```

```python
# BFS preorder traversal
# each node has an interval. pass in an interval when iterating down
# when traversal left node, change the right edge of the interval to the node value
# when traversal right node, change the left edge of the interval to the node value 
def isValidBST(root, left = float('-inf'), right = float('inf')):
    if root is None:
        return True
    x = root.val
    return left < x < right and isValidBST(root.left, left, x) and isValidBST(root.right, x, right)
```

```python
# BFS inorder traversal
# traverse left tree first and then right tree
# determine whether the node is greater than its previous one
pre = float('-inf')
def isValidBST(root):
    if root is None:
        return True 
    if not isValidBST(root.left):
        return False
    if root.val <= pre:
        return False
    pre = root.val
    return isValidBST(root.right)
```

```python
# BFS postorder traversal
def isValidBST(root) -> bool:
    # return the min and max element of the tree begin by this node
    def f(node):
        # to skip the following if statement
        if node is None:
            return inf, -inf
        l_min, l_max = f(node.left)
        r_min, r_max = f(node.right)
        x = node.val
        # when it is not a BST
        if x <= l_max or x >= r_min:
            return -inf, inf
        # return l_min, r_max
        # consider the child node is empty, l_min is inf and r_max is -inf
        return min(l_min, x), max(r_max, x)
    # if it is not a BST, it will return (-inf, inf)
    return f(root)[0] != -inf
```
