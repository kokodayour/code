```python
def merge(intervals: list[list[int]]) -> list[list[int]]:
    ans = []
    intervals.sort(key=lambda a:a[0])
    L = intervals[0][0]
    R = intervals[0][1]
    for i in intervals[1:]:
        if i[0] <= R and i[1] >=R:
            R = i[1]
        if i[0] > R:
            ans.append([L,R])
            L = i[0]
            R = i[1]
    return ans+[[L,R]]
```

```python
def merge(intervals: list[list[int]]) -> list[list[int]]:
    intervals.sort(key=lambda a:a[0])
    ans = [intervals[0]]
    for i in intervals[1:]:
        if i[0] <= ans[-1][1]:
            ans[-1][1] = max(ans[-1][1], i[1])
        else:
            ans.append(i)
    return ans
```

```python
intervals = [[1,3],[2,6],[8,10],[15,18]]
intervals = [[1,4],[0,4]]
merge(intervals)
```
