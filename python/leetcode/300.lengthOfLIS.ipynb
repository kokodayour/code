```python
# 单调栈 每次要踢元素的时候统计栈内的元素个数
def lengthOfLIS(nums: list[int]) -> int:
    st = [nums[0]]
    ans = 0
    for x in nums[1:]:
        if x in st:
            continue
        if x < st[-1]:
            ans = max(len(st), ans)
        while st and x < st[-1]:
            st.pop()
        st.append(x)
    return max(ans, len(st))
```

```python
nums = [10,9,2,5,3,7,101,18]
nums = [0,1,0,3,2,3]
lengthOfLIS(nums)
```
