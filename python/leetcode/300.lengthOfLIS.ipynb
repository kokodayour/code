```python
# 暴力+递归
# 递归函数: 求以索引i开始的最长子序列
class Solution:
    def lengthOfLIS(self, nums: List[int]) -> int:
        self.nums = nums  # 存储为实例变量，供递归函数使用
        return max(self._func(i) for i in range(len(nums)))
    
    def _func(self, i):
        """辅助递归函数，计算以 nums[i] 开头的最长递增子序列长度"""
        if i == len(self.nums) - 1:
            return 1
        
        max_len = 1
        for j in range(i + 1, len(self.nums)):
            if self.nums[j] > self.nums[i]:
                max_len = max(max_len, self._func(j) + 1)
        return max_len
```

```python
# 记忆化搜索 空间换时间
class Solution:
    def lengthOfLIS(self, nums: List[int]) -> int:
        self.nums = nums
        self.lookup = {}  # 实例级缓存，避免全局污染
        return max(self._func(i) for i in range(len(nums)))
    
    def _func(self, i):
        """计算以 nums[i] 开头的最长递增子序列长度"""
        if i in self.lookup:
            return self.lookup[i]

        if i == len(self.nums) - 1:
            return 1
        
        max_len = 1
        for j in range(i + 1, len(self.nums)):
            if self.nums[j] > self.nums[i]:
                max_len = max(max_len, self._func(j) + 1)
        
        self.lookup[i] = max_len
        return max_len
```

```python
# 改写成迭代模式
def lengthOfLIS(nums: List[int]):
    n = len(nums)
    dp = [1]*n
    # 外循环迭代dp 内循环迭代max(dp[4],dp[3],dp[2])
    for i in reversed(range(n)):
        for j in range(i+1, n):
            if nums[j] > nums[i]:
                dp[i] = max(dp[i], dp[j]+1)
    return max(dp)
```

```python
# 使用注解
from functools import cache
def lengthOfLIS(nums):
    @cache
    def dp(i):
        ans = 0
        for j in range(i):
            if nums[j] < nums[i]:
                ans = max(ans, dp(j))
        return ans+1
    return max(dp(i) for i in range(len(nums)))
```

```python
# 贪心+二分查找
# dp[i]是末尾元素为nums[i]的LIS长度
# 反过来定义f[i] 表示长度为i+1的LIS末尾元素的最小值
import bisect
def lengthOfLIS(nums):
    f = []
    for x in nums:
        j = bisect.bisect_left(f,x)
        if j == len(f):
            f.append(x)
        else:
            f[j] = x
    return len(f)
```
