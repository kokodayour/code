```python
def maxPathSum(root):
    ans = root.val
    # 返回该节点为头节点的最大路径和
    def dfs(root):
        nonlocal ans
        if root is None:
            return 0
        left = dfs(root.left)
        right = dfs(root.right)
        left1 = left + root.val
        right1 = right + root.val
        # 最大值可能有5种情况 之前最大值, 左+根, 右+根, 根, 左+根+右
        ans = max(ans, left1, right1, root.val, left1+right)
        # 节点的最大路径和可能有三种种情况 左+根, 右+根, 根
        return max(left1, right1, root.val)
    dfs(root)
    return ans
```

```python
# 修改版 如果该节点为头节点的最大路径和是负数, 将其置为0避免讨论
def maxPathSum(root):
    ans = root.val
    # 返回该节点为头节点的最大路径和
    def dfs(root):
        nonlocal ans
        if root is None:
            return 0
        left = dfs(root.left)
        right = dfs(root.right)
        # 更新答案
        ans = max(left+root.val+right, ans)
        return max(left+root.val, right+root.val, 0)
    dfs(root)
    return ans
```
