```python
def sortedArrayToBST(nums):
    if not nums:
        return None
    # get centre node
    m = len(nums)//2
    # recursively call function for left and right subtree respectively
    left = sortedArrayToBST(nums[:m])
    right = sortedArrayToBST(nums[m+1:])
    return TreeNode(nums[m], left, right)
```
