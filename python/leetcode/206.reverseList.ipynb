```python
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next

def reverseList(head: Optional[ListNode]) -> Optional[ListNode]:
    a = []
    p = head
    while p:
        a.append(p.val)
        p = p.next
    a.reverse()
    q = head
    for x in a:
        q.val = x
        q = q.next
    return head
```

```python
# 头插法
def reverseList(head: Optional[ListNode]) -> Optional[ListNode]:
    pre = None
    cur = head
    while cur:
        nxt = cur.next
        cur.nxt = pre
        pre = cur
        cur = nxt
    return pre
```
