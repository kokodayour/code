```python
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next

def reverseList(head: Optional[ListNode]) -> Optional[ListNode]:
    a = []
    p = head
    while p:
        a.append(p.val)
        p = p.next
    a.reverse()
    q = head
    for x in a:
        q.val = x
        q = q.next
    return head
```

```python
# 头插法 相当于将链表中的元素不断地加到pre前面
def reverseList(head: Optional[ListNode]) -> Optional[ListNode]:
    pre = None
    cur = head
    while cur:
        nxt = cur.next
        cur.nxt = pre
        pre = cur
        cur = nxt
    return pre
```

```python
# 头插法的另外一种写法 相当于不断地把后面的数转到前面来
def reverseList(head: Optional[ListNode]) -> Optional[ListNode]:
    dummy = ListNode(next=head)
    cur = head
    # 之所以判断cur是因为cur可能为None而不具有next属性
    while cur and cur.next:
        next = cur.next
        cur.next = next.next
        next.next = dummy.next
        dummy.next = next
    return dummy.next
```
