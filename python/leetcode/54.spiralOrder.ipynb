```python
DIRS = (0,1), (1,0), (0,-1),(-1,0)
def spiralOrder(matrix: list[list[int]]) -> list[int]:
    m, n = len(matrix), len(matrix[0])
    ans = []
    # di表示当前方向
    i = j = di = 0
    for _ in range(m*n):
        ans.append(matrix[i][j])
        matrix[i][j] = None # 标记, 表示已访问
        x, y = i + DIRS[di][0], j + DIRS[di][1] # 下一步的位置
        # 如果(x, y)出界或者已经访问过
        if x < 0 or x >= m or y < 0 or y >= n or matrix[x][y] is None:
            di = (di+1)%4 # 右转90°
        i += DIRS[di][0]
        j += DIRS[di][1]
    return ans
```

```python
DIRS = (0,1), (1,0), (0,-1),(-1,0)
def spiralOrder(matrix: list[list[int]]) -> list[int]:
    m, n = len(matrix), len(matrix[0])
    size = m*n
    ans = []
    i, j, di = 0, -1, 0
    while len(ans) < size:
        dx, dy = DIRS[di]
        for _ in range(n):
            i += dx
            j += dy
            ans.append(matrix[i][j])
        di = (di+1)%4
        n, m = m-1, n
    return ans
```

```python
matrix = [[1,2,3],[4,5,6],[7,8,9]]
spiralOrder(matrix)
```
