```python
# 滑动窗口
def summaryRanges(nums):
    ans = []
    L = 0
    n = len(nums)
    for R in range(n):
        if R + 1 < n and nums[R+1]-nums[R]==1:
            continue
        else:
            if L != R:
                ans.append(f"{nums[L]}->{nums[R]}")
            else:
                ans.append(f"{nums[R]}")
            L = R+1
    return ans
```

```python
nums = [0,1,2,4,5,7]
summaryRanges(nums)
```
