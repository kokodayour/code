```python
# 错误解法: 变长滑动窗口
# 当窗口中的元素和小于阈值的时候想当然地移动窗口右边界
# 但由于窗口中的元素可能小于0 所以还可以移动窗口的左边界
# 总之无法确定怎么移动窗口
def subarraySum(nums: list[int], k: int) -> int:
    L = R = ans = temp = 0
    n = len(nums)
    while R < n:
        temp += nums[R]
        while temp > k:
            if L >= R:
                break
            temp -= nums[L]
            L += 1
        if temp == k:
            ans += 1
        R += 1
    return ans
```

```python
# 采用前缀和解法
# 问题转换成: 求前缀和数组中两个元素的差值等于阈值的元素对个数 或者单元素的值就是阈值的个数
def subarraySum(nums: list[int], k: int) -> int:
    pre_sum = []
    total = 0
    ans = 0
    freq = {}
    for x in nums:
        total += x
        pre_sum.append(total)
    for x in pre_sum:
        if x == k:
            ans += 1
        if x - k in freq.keys():
            ans += freq[x-k]
        freq[x] = freq.get(x, 0) + 1
    return ans
```

```python
# 一边计算前缀和 一边遍历前缀和
def subarraySum(self, nums: list[int], k: int) -> int:
    ans = s = 0
    cnt = defaultdict(int)
    cnt[0] = 1  # s[0]=0 单独统计 避免了单元素的讨论
    for x in nums:
        s += x
        ans += cnt[s - k]
        cnt[s] += 1
    return ans
```

```python
nums = [1,2,3]
k = 3
subarraySum(nums, k)
```
