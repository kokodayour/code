```python
# Error solution: variable lenght sliding window
# when the sum of the elems in the window is less than threshold, move the right edge of this window
# However, since the elems in the window may be negatives, it can also move the left edge
# Anyway, i can't determine how to move the window
def subarraySum(nums: list[int], k: int) -> int:
    L = R = ans = temp = 0
    n = len(nums)
    while R < n:
        temp += nums[R]
        while temp > k:
            if L >= R:
                break
            temp -= nums[L]
            L += 1
        if temp == k:
            ans += 1
        R += 1
    return ans
```

```python
# prefix-sum solution
# The problem is converted into: get the number of elem-pairs in the prefix-sum array that have a different equal to the threshold
# or the number of elems which the value of one of them is equal to threshold
def subarraySum(nums: list[int], k: int) -> int:
    pre_sum = []    
    total = 0
    ans = 0
    freq = {}
    for x in nums:
        total += x
        pre_sum.append(total)
    for x in pre_sum:
        if x == k:
            ans += 1
        if x - k in freq.keys():
            ans += freq[x-k]
        freq[x] = freq.get(x, 0) + 1
    return ans
```

```python
# caculate and traverse pre-sum
def subarraySum(self, nums: list[int], k: int) -> int:
    ans = s = 0
    cnt = defaultdict(int)
    cnt[0] = 1  # s[0]=0 单独统计 避免了单元素的讨论
    for x in nums:
        s += x
        ans += cnt[s - k]
        cnt[s] += 1
    return ans
```
