```python
def reverseBetween(head, left, right):
    # dummy: record head node
    # cur: current node
    dummy = cur = ListNode(next = head)
    for _ in range(left-1):
        cur = cur.next
    # cur is now the pre node of left
    # reverse Linkedlist
    old = cur.next
    new = ListNode()
    for _ in range(right-left+1):
        temp = old.next
        old.next = new
        new = old
        old = temp
    # old has been the node of right
    cur.next.next = old
    cur.next = new
    return dummy.next
```

```python
def reverseBetween(head, left: int, right: int):
    pre = dummy = ListNode(next = head)
    for _ in range(left-1):
        pre = pre.next
    cur = pre.next
    # each cycle will put `nxt` to the left
    for _ in range(right-left):
        nxt = cur.next
        cur.next = nxt.next
        nxt.next = pre.next
        pre.next = nxt
    return dummy.next
```
