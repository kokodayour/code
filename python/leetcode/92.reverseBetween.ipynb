```python
from node import *
def reverseBetween(head, left: int, right: int):
    # pre和suf是要交换的字符串的前驱和后继
    pre = suf = dummy = ListNode(next=head)
    for _ in range(left-1):
        pre = pre.next
    for _ in range(right+1):
        suf = suf.next
    cur = pre.next
    for _ in range(right-left+1):
        nxt = cur.next
        cur.next = suf
        suf = cur
        cur = nxt
    pre.next = suf
    return dummy.next
```

```python
def reverseBetween(head, left: int, right: int):
    pre = dummy = ListNode(next = head)
    for _ in range(left-1):
        pre = pre.next
    cur = pre.next
    for _ in range(right-left):
        next = cur.next
        cur.next = next.next
        next.next = pre.next
        pre.next = next
    return dummy.next
```
