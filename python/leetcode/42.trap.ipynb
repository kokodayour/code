```python
# 以每一个柱子为中心，向两边扩展，找到左右两边的最高柱子，取最小值减去当前柱子的高度
def trap(height: list[int]) -> int:
    n = len(height)
    pre = [0] * n
    pre[0] = height[0]
    for i in range(1, n):
        pre[i] = max(pre[i - 1], height[i])
    suf = [0] * n
    suf[-1] = height[-1]
    for i in range(n - 2, -1, -1):
        suf[i] = max(suf[i + 1], height[i])
    ans = 0
    for pi, fi ,i in zip(pre, suf, height):
        ans += min(pi, fi) - i
    return ans
```

```python
# 双指针解法 无需存储前缀后缀数组
def trap(height: list[int]) -> int:
    n = len(height)
    pre = 0
    suf = 0
    L = 0
    R = n - 1
    ans = 0
    while R >= L:
        pre = max(height[L], pre)
        suf = max(height[R], suf)
        if pre < suf:
            ans += pre - height[L]
            L += 1
        else:
            ans += suf - height[R]
            R -= 1
    return ans
```

```python
# 单调栈解法 按行存储雨水
def trap(height: list[int]) -> int:
    st = []
    ans = 0
    for i, h in enumerate(height):
        while st and h >= height[st[-1]]:
            bottom_h = height[st.pop()]
            if not st:
                break
            left = st[-1]
            dh = min(height[left], h) - bottom_h
            ans += dh * (i - left - 1)
        st.append(i)
    return ans
```

```python
height = [0,1,0,2,1,0,1,3,2,1,2,1]
trap(height)
```
