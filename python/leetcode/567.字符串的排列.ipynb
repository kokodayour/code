```python
from collections import Counter
def checkInclusion(s1: str, s2: str) -> bool:
    n = len(s1)
    freq = Counter(s1)
    temp = Counter(s2[:n])
    if temp == freq:
        return True
    for in_, out in zip(s2[n:], s2):
        temp[in_] += 1
        temp[out] -= 1
        if temp == freq:
            return True
    return False
```

```python
s1 = "ab" 
s2 = "eidbaooo"
checkInclusion(s1, s2)
```
