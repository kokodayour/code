```python
def buildTree(preorder, inorder):
    """
    return a tree's root based on known preorder and inorder
    """
    # preorder can be empty when iteration
    if preorder == []:
        return
    idx = inorder.index(preorder[0])
    left = buildTree(preorder[1:1+idx], inorder[:idx])
    right = buildTree(preorder[1+idx:], inorder[1+idx:])
    return buildTree(TreeNode(preorder[0], left, right))
```

```python
def buildTree(preorder, inorder):
    lookup = {val: idx for idx, val in enumerate(inorder)}
    # change the params to the left and right endpoints of subarrays so as to aovid copying them  
    def dfs(pre_l, pre_r, in_l, in_r):
        if pre_l == pre_r: # 空节点
            return None
        idx = lookup[preorder[pre_l]] - in_l
        # use index to aovid slice
        left = dfs(1+pre_l, 1+pre_l+idx, in_l, in_l+idx)
        right = dfs(1+pre_l+idx, pre_r, in_l+idx+1, in_r)
        return TreeNode(preorder[pre_l], left, right)
    return dfs(0, len(preorder), 0, len(inorder))
```
