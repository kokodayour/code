```python
def subsetWithDup(nums):
    ans = [[]]
    temp = []
    n = len(nums)
    def dfs(i):
        if i<n:
            # not choose nums[i]
            dfs(i+1)
            # choose nums[i]
            temp.append(nums[i])
            ans.append(temp.copy())
            dfs(i+1)
    dfs(0)
    return ans
```

```python
def subsetWithDup(nums):
    ans = [[]]
    temp = []
    n = len(nums)
    def dfs(i, nums):
        if i<n:
            for x in nums:
                temp.append(x)
                ans.append(temp)
                dfs(i, nums.remove(x))
    dfs(0, nums)
    return ans
```

```python
subsetWithDup([1,2,2])
```
