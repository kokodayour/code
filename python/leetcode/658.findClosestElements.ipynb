```python
import numpy as np


def gcn_layer(A, X, W):
    """
    实现一个简单的 GCN 层
    :param A: 邻接矩阵
    :param X: 节点特征矩阵
    :param W: 权重矩阵
    :return: 经过 GCN 层处理后的节点特征矩阵
    """
    # 计算度矩阵
    D = np.diag(np.sum(A, axis=1))

    # 计算归一化邻接矩阵
    D_inv_sqrt = np.linalg.inv(np.sqrt(D))
    A_hat = D_inv_sqrt @ A @ D_inv_sqrt
    # 计算经过 GCN 层的输出
    Z = np.dot(np.dot(A_hat, X), W)

    # 使用 ReLU 激活函数
    output = np.maximum(0, Z)
    return output


# 示例数据
# 邻接矩阵
A = np.array([
    [0, 1, 1],
    [1, 0, 1],
    [1, 1, 0]
])

# 节点特征矩阵
X = np.array([
    [1, 0],
    [0, 1],
    [1, 1]
])

# 权重矩阵
input_dim = X.shape[1]
output_dim = 2
W = np.random.randn(input_dim, output_dim)

# 前向传播
output = gcn_layer(A, X, W)
print("经过 GCN 层处理后的节点特征矩阵:")
print(output)
    
```
