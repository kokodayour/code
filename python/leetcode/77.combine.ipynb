```python
# the width and dipth of tree is n and k respectively
def combine(n, k):
    path = [0] * k
    ans = []

    def dfs(start, i):
        if i == k:
            ans.append(path.copy())
            return
        # without pruning, as long as x is within the range, it will recurse
        for x in range(start+1, n+1):
            path[i] = x
            dfs(x, i+1)

    dfs(0, 0)
    return ans
```

```python
# iterate which number to choose next
def combine(n, k):
    path = [0]*k
    ans = []

    def dfs(start, i):
        if i == k:
            ans.append(path.copy())
            return
        # the remain number of digitals to be selected
        remain = k - i
        # prun: x cannot greater than , otherwise the remaining digitals are insufficient
        MAX = n - (remain - 1)
        for x in range(start+1, MAX+1):
            path[i] = x
            dfs(x, i+1) 

    dfs(0, 0)
    return ans
```

```python
# choose or not
def combine(n, k):
    ans = []
    path = []

    def dfs(i):
        # the remain number of digitals to be selected
        remain = k - len(path)
        if remain == 0:
            ans.append(path.copy())
            return
        # not choose i and there are enough digitals to be selected
        if i > remain:
            dfs(i-1)
        # choose i
        path.append(i)
        dfs(i-1)
        # backtrack
        path.pop()

    dfs(n)
    return ans
```
