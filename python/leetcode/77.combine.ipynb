```python
# the width and dipth of tree is n and k respectively
def combine(n, k):
    path = [0] * k
    ans = []

    def dfs(start, i):
        if i == k:
            ans.append(path.copy())
            return
        for x in range(start+1, n+1):
            path[i] = x
            dfs(x, i+1)

    dfs(0, 0)
    return ans
```

```python
combine(4,2)
```
