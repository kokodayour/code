```python
def orangesRotting(grid):
    m, n = len(grid), len(grid[0])
    # count the number of fresh oranges
    fresh = 0
    # record positions of rotten oranges
    q = []
    for i, row in enumerate(grid):
        for j, x in enumerate(row):
            if x == 1:
                fresh += 1
            if x == 2:
                q.append((i, j))
    ans = 0
    while q and fresh:
        ans += 1 # after one minute
        tmp = q
        # clean up rotten oranges array to aovid subsequently repeated traversal 
        q = []
        for x, y in tmp:
            for i, j in (x-1, y), (x+1, y), (x, y-1), (x, y+1): # four directions
                if (0 <= i < m) and (0 <= j < n) and grid[i][j] == 1:
                    fresh -= 1
                    grid[i][j] = 2
                    q.append((i, j))
    return -1 if fresh else ans
```
