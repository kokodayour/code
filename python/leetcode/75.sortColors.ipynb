```python
def sortColors(nums) -> None:
    l0 = l1 = 0
    for x in nums:
        if x == 0:
            l0 += 1
        elif x == 1:
            l1 += 1
    for i in range(len(nums)):
        if l0:
            nums[i] = 0
            l0 -= 1
        elif l1:
            nums[i] = 1
            l1 -= 1
        else:
            nums[i] = 2
    return nums
```

```python
# 先遍历一次 将0与其它数交换 然后再遍历一次 将1与2交换
def sortColors(nums):
    # 先指向头部0的位置
    ptr = 0
    n = len(nums)
    for i in range(n):
        if nums[i] == 0:
            nums[i], nums[ptr] = nums[ptr], nums[i]
            ptr += 1
    for i in range(ptr, n):
        if nums[i] == 1:
            nums[i], nums[ptr] = nums[ptr], nums[i]
            ptr += 1
    return nums
```

```python
# p0记录头部0的位置 p1记录头部1的位置
# 当正在遍历的元素小于2时 移动p0和p1的位置 并修改对应位置的元素
def sortColors(nums):
    p0 = p1 = 0
    for i, x in enumerate(nums):
        nums[i] = 2
        if x <= 1:
            nums[p1] = 1
            p1 += 1
        if x == 0:
            nums[p0] = 0
            p0 += 1
    return nums
```

```python
def sortColors(nums):
    n = len(nums)
    p0 = 0
    p2 = n-1
    i = 0
    while i <= p2:
        while i <= p2 and nums[i] == 2:
            nums[i], nums[p2] = nums[p2], nums[i]
            p2 -= 1
        if nums[i] == 0:
            nums[i], nums[p0] = nums[p0], nums[i]
            p0 += 1
        i += 1
```

```python
nums = [2,0,2,1,1,0]
sortColors(nums)
```
