```python
def sortColors(nums) -> None:
    l0 = l1 = 0
    for x in nums:
        if x == 0:
            l0 += 1
        elif x == 1:
            l1 += 1
    for i in range(len(nums)):
        if l0:
            nums[i] = 0
            l0 -= 1
        elif l1:
            nums[i] = 1
            l1 -= 1
        else:
            nums[i] = 2
    return nums
```

```python
# 先遍历一次 将0与其它数交换 然后再遍历一次 将1与2交换
def sortColors(nums):
    # 先指向头部0的位置
    ptr = 0
    n = len(nums)
    for i in range(n):
        if nums[i] == 0:
            nums[i], nums[ptr] = nums[ptr], nums[i]
            ptr += 1
    for i in range(ptr, n):
        if nums[i] == 1:
            nums[i], nums[ptr] = nums[ptr], nums[i]
            ptr += 1
    return nums
```

```python
nums = [2,0,2,1,1,0]
sortColors(nums)
```
