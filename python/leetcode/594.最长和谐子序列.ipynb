```python
from collections import Counter
def findLHS(nums: list[int]) -> int:    
    # 统计每个数字出现的频率
    freq = Counter(nums)
    # 初始化最大长度为 0
    max_length = 0
    # 遍历 Counter 中的每个元素
    for num in freq:
        # 检查 num + 1 是否也在 Counter 中
        if num + 1 in freq:
            # 计算当前和谐子序列的长度
            current_length = freq[num] + freq[num + 1]
            # 更新最大长度
            max_length = max(max_length, current_length)
    return max_length
```

```python
def findLHS(nums: list[int]) -> int:
    cnt = Counter(nums)
    return max((val + cnt[key + 1] for key, val in cnt.items() if key + 1 in cnt), default=0)
```

```python
nums = [1,3,2,2,5,2,3,7]
# nums = [1,1,1,1]
findLHS(nums)
```
