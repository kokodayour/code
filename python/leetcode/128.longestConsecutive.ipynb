```python
def longestConsecutive(nums: list[int]) -> int:
    nums = sorted(list(set(nums)))
    ans = temp = 1
    for x in nums:
        if x+1 not in nums:
            ans = max(ans, temp)
            temp = 1
        else:
            temp += 1
    return 0 if nums == [] else ans
```

```python
# 1.find the start
# 2.get its seq
def longestConsecutive(nums: list[int]) -> int:
    # 将列表转换为集合，方便快速查找元素
    st = set(nums)
    longest_seq = 0

    for num in st:
        # 判断当前元素是否是连续序列的起始元素
        if num - 1 not in st:
            current_num = num
            current_seq = 1

            # 不断检查下一个连续的数字是否存在于集合中
            while current_num + 1 in st:
                current_num += 1
                current_seq += 1

            # 更新最长连续序列的长度
            longest_seq = max(longest_seq, current_seq)

    return longest_seq
```

```python
nums = [0,3,7,2,5,8,4,6,0,1]
nums = [0,-1]
longestConsecutive(nums)
```
