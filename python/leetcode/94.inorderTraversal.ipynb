```python
class Solution:
    def __init__(self):
        self.ans = []  # 实例变量，每次创建 Solution 实例时初始化

    # 递归
    def inorderTraversal(self, root: Optional[TreeNode]) -> List[int]:
        if root:
            self.inorderTraversal(root.left)
            self.ans.append(root.val)
            self.inorderTraversal(root.right)
        return self.ans
```

```python
# 递归隐式地维护了一个栈 迭代显式维护一个栈
def inorderTraversal(root):
    st = []
    ans = []
    while root or st:
        while root:
            st.append(root)
            root = root.left
        root = st.pop()
        ans.append(root.val)
        root = root.right
    return ans
```

```python
# 莫里斯遍历
def inorderTraversal(root):
    ans = []
    cur = root
    while cur:
        if not cur.left: # 情况1: 无左子树
            ans.append(cur.val)
            cur = cur.right
        else: # 情况2: 有左子树
        # 找到中序前驱节点(左子树的最右节点)
            pre = cur.left
            while pre.right and pre.right != cur:
                pre = pre.right
            if not pre.right: # 建立线索
                pre.right = cur
                cur = cur.left
            else: # 说明已建立线索 断开线索
                pre.right = None
                ans.append(cur.val)
                cur = cur.right
    return ans
```
