```python
class Node:
    def __init__(self, x: int, next: 'Node' = None, random: 'Node' = None):
        self.val = int(x)
        self.next = next
        self.random = random

def copyRandomList(head:'Node'):
    if not head:
        return None
    # l1和l2用于存储原链表和新链表
    # 列表里添加None 防止后续索引random的时候报错
    l1 = [None]
    l2 = [None]
    dummy = cur = Node(-1)
    while head:
        l1.append(head)
        l2.append(Node(x=head.val))
        cur.next = l2[-1]
        cur = cur.next
        head = head.next
    head1 = l1[1]
    head2 = l2[1]
    while head1:
        i = l1.index(head1.random)
        head2.random = l2[i]
        head1 = head1.next
        head2 = head2.next
    return dummy.next
```

```python
# 将新旧链表进行交错1->1'->2->2'
# 1指向random 1'只需要指向random的下一个节点
# 最后将1'->2'提取出来
def copyRandomList(head:'Node'):
    if not head:
        return None
    
    # 形成交错链表
    cur = head
    while cur:
        cur.next = Node(x=cur.val, next=cur.next)
        cur = cur.next.next
    
    # 遍历交错链表 设定random
    cur = head
    while cur:
        # cur.next.random = cur.random.next if cur.random else None
        # cur.random 默认是None
        if cur.random:
            cur.next.random = cur.random.next
        cur = cur.next.next
    
    # 取出新链表
    cur = head.next
    while cur.next:
        cur.next = cur.next.next
        cur = cur.next
    
    return head.next
```
