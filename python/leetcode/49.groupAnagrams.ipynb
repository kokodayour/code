```python
# dict is not required
from collections import Counter
def groupAnagrams(strs: list[str]) -> list[list[str]]:
    freq = {}
    for x in strs:
        cnt = tuple(sorted(Counter(x).items()))
        freq[cnt] = freq.get(cnt, []) + [x]
    return [x for x in freq.values()]
```

```python
def groupAnagrams(strs: list[str]) -> list[list[str]]:
    lookup = {}
    for x in strs:
        sorted_x = tuple(sorted(x))
        lookup[sorted_x] = lookup.get(sorted_x, []) + [x]
    return list(lookup.values())
```

```python
strs = ["eat", "tea", "tan", "ate", "nat", "bat"]
groupAnagrams(strs)
```
