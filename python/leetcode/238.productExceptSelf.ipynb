```python
# 前缀乘积与后缀成绩
def productExceptSelf(nums: list[int]) -> list[int]:
    n = len(nums)
    pre = [1] * n
    suf = [1] * n
    for i in range(1,n):
        pre[i] = pre[i-1]*nums[i-1]
    for i in range(n-2,-1,-1):
        suf[i] = suf[i+1]*nums[i+1]
    return [p*s for p, s in zip(pre, suf)]
```

```python
def productExceptSelf(nums: list[int]) -> list[int]:
    n = len(nums)
    suf = [1] * n
    for i in range(n-2,-1,-1):
        suf[i] = suf[i+1]*nums[i+1]
    # 在遍历的过程中计算pre
    pre = 1
    for i,x in enumerate(nums):
        suf[i] *= pre
        pre *= x
    return suf
```
