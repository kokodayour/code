```python
def maxDepth(root):
    if root is None:
        return 0
    return max(maxDepth(root.left), maxDepth(root.right)) + 1
```

```python
def maxDepth(root):
    ans = 0
    def f(node, cnt):
        if node is None:
            return
        cnt += 1
        nonlocal ans
        ans = max(ans, cnt)
        f(node.left, cnt)
        f(node.right, cnt)
    f(root, 0)
    return ans
```
