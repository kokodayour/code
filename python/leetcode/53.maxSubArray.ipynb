```python
# 问题转换成：求解前缀和数组两个元素之差的最大值

from itertools import accumulate
def maxSubArray(nums: list[int]) -> int:
    pre_sum = list(accumulate(nums, initial=0))
    min_val = pre_sum[0]
    ans = pre_sum[1]
    for ptr in pre_sum[1:]:
        # 更新最大子数组和
        ans = max(ans, ptr - min_val)
        # 更新最小前缀和
        min_val = min(min_val, ptr)
    return ans
```

```python
def maxSubArray(nums: list[int]) -> int:
    minimum = pre_sum = 0
    ans = nums[0]
    for x in nums:
        pre_sum += x
        ans = max(ans, pre_sum-minimum)
        minimum = min(minimum, pre_sum)
    return ans
```

```python
# 问题转换成(动态规划): 求解以元素i结尾的最大子数组和dp[i]
# 状态转移方程: dp[i] = max(dp[i-1], 0) + nums[i]
def maxSubArray(nums):
    n = len(nums)
    dp = n*[0]
    dp[0] = nums[0]
    for i in range(1,n):
        dp[i] = max(dp[i-1]+nums[i], nums[i])
    return max(dp)
```

```python
# 优化空间复杂度
def maxSubArray(nums):
    ans = dp = nums[0]
    for x in nums[1:]:
        dp = max(x,dp+x)
        ans = max(ans, dp)
    return ans
```
