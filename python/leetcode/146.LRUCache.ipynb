```python
# 不能使用pop和remove方法 它们的时间复杂度都是On
class LRUCache:

    def __init__(self, capacity: int):
        self.capacity = capacity
        # 用于索引(key, value)
        self.lookup = {}
        # 按时间顺序创建列表 列表元素是key
        self.l = []

    def get(self, key: int) -> int:
        ans = self.lookup.get(key, -1)
        if ans != -1:
            self.l.remove(key)
            self.l.append(key)
        return ans
        

    def put(self, key: int, value: int) -> None:
        if self.get(key) != -1:
            self.l.remove(key)
        if len(self.l) == self.capacity:
            self.l.pop(0)
        self.lookup[key] = value
        self.l.append(key)
```

```python
class Node:
    # 提高访问属性的速度 并节省内存
    __slots__ = 'pre', 'next', 'key', 'value'

    def __init__(self, key=0, value=0) -> None:
        self.key = key
        self.value = value

class LRUCache:
    def __init__(self, capacity: int):
        self.capacity = capacity
        self.dummy = Node() # 哨兵节点
        self.dummy.pre = self.dummy
        self.dummy.next = self.dummy
        self.lookup = {} # {key:Node}
    def get_node(self, key):
        if key not in self.lookup:
            return None
        node = self.lookup[key]
        self.remove(node)
        self.push_front(node)
        return node
    def get(self, key: int) -> int:
        node = self.get_node(key)
        return node.value if node else -1
    def put(self, key: int, value: int) -> None:
        node = self.get_node(key)
        if node:
            node.value = value
            return
        self.lookup[key] = node = Node(key, value)
        self.push_front(node)
        if len(self.lookup) > self.capacity:
            back_node = self.dummy.pre
            del self.lookup[back_node.key]
            self.remove(back_node)
    # 删除一个节点
    def remove(self, x:Node):
        x.pre.next = x.next
        x.next.pre = x.pre
    def push_front(self, x:Node):
        x.pre = self.dummy
        x.next = self.dummy.next
        x.pre.next = x
        x.next.pre = x
```
