```python
# 不能使用pop和remove方法 它们的时间复杂度都是On
class LRUCache:

    def __init__(self, capacity: int):
        self.capacity = capacity
        # 用于索引(key, value)
        self.lookup = {}
        # 按时间顺序创建列表 列表元素是key
        self.l = []

    def get(self, key: int) -> int:
        ans = self.lookup.get(key, -1)
        if ans != -1:
            self.l.remove(key)
            self.l.append(key)
        return ans
        

    def put(self, key: int, value: int) -> None:
        if self.get(key) != -1:
            self.l.remove(key)
        if len(self.l) == self.capacity:
            self.l.pop(0)
        self.lookup[key] = value
        self.l.append(key)
```

```python
class LRUCache:

    def __init__(self, capacity: int):
        self.capacity = capacity
        # 用于索引(key, 双向链表的结点(value, pre, next)) 按时间顺序组件双向链表
        self.lookup = {}
        # 链表的尾部节点
        self.tail = (-1, (-1,-1,-1))


    def get(self, key: int) -> int:
        ans = self.lookup.get(key, -1)
        if ans != -1:
            self.lookup[self.lookup[key][1]][2] = self.lookup[key][2]
            self.tail[1][2] = self.lookup[key]
            self.lookup[key][1] = self.tail[0]
            self.lookup[key][2] = None
            self.tail = (key, self.lookup[key])
        return ans
        

    def put(self, key: int, value: int) -> None:
        if len(self.lookup) == self.capacity:
            pass
        else:
            self.lookup[key] = (value, self.tail[0], -1)
            self.tail = (key, (value, self.tail[0], -1))
```
