```python
def minDeletion(s, k):
    lookup={}
    for x in s:
        lookup[x] = lookup.get(x,0) + 1
    values = list(lookup.values())
    values.sort()
    if k >= len(values):
        return 0
    return sum(values[:len(values) - k])
```

```python
s = 'abc'
k = 2
s = 'aabb'
k = 1
minDeletion(s, k)
```
