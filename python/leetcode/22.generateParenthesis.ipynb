```python
def generateParenthesis(n: int) -> list[str]:
    ans = []
    path = ['']*(n*2)  # 存储当前路径
    
    def dfs(left, right):
        if right == n:
            ans.append(''.join(path))
            return
        # 可以填左括号
        if left < n:
            path[left+right] = '('
            dfs(left+1, right)
        # 可以填右括号
        if right < left:
            path[left+right] = ')'
            dfs(left, right+1)
    
    dfs(0, 0)
    return ans
```

```python
def generateParenthesis(n: int) -> list[str]:
    ans = []
    left = right = n
    st = []  # 栈用于验证括号匹配
    path = []  # 存储当前路径
    
    def dfs(choice):
        nonlocal left, right
        
        # 执行当前选择
        if choice == 'left':
            if left == 0:  # 没有左括号可用了
                return
            st.append('(')  # 压入左括号到栈
            path.append('(')
            left -= 1
        else:  # choice == 'right'
            if right == 0 or not st:  # 没有右括号或栈空（无法匹配）
                return
            st.pop()  # 弹出匹配的左括号
            path.append(')')
            right -= 1
        
        # 检查是否完成
        if left == 0 and right == 0 and not st:
            ans.append(''.join(path))
        else:
            # 尝试下一步的所有可能选择
            dfs('left')   # 尝试左括号
            dfs('right')  # 尝试右括号
        
        # 回溯：撤销当前选择
        if choice == 'left':
            path.pop()
            st.pop()  # 弹出之前压入的左括号
            left += 1
        else:  # choice == 'right'  
            path.pop()
            st.append('(')  # 恢复被pop掉的左括号
            right += 1
    
    dfs('left')  # 从左括号开始
    return ans
```

```python
# 枚举下一个左括号的位置
def generateParenthesis(n):
    ans = []
    # 记录左括号的下标
    path = []

    # 目前填了i个括号
    # balance = 这i个括号中的左括号个数-右括号个数
    def dfs(i, balance):
        if len(path) == n:
            s = [')']*(n*2)
            for j in path:
                s[j] = '('
            ans.append(''.join(s))
            return
        # 在填下一个左括号之前可以填入0-balance个右括号 这样就可以得到下一个左括号的位置
        for right in range(balance+1):
            # 先填right个右括号 然后填1个左括号 记录左括号的下标i+right
            path.append(i+right)
            dfs(i+right+1, balance-right+1)
            path.pop()

    dfs(0, 0)
    return ans
```
