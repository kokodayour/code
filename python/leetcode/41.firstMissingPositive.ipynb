```python
def firstMissingPositive(nums: list[int]) -> int:
    i = 1
    while True:
        if i in nums:
            i += 1
            continue
        else:
            return i
```

```python
def firstMissingPositive(nums: list[int]) -> int:
    n = len(nums)
    for i in range(n):
        # 如果当前学生的学号在[1,n]中, 但真身没有坐在正确的座位上
        while 1 <= nums[i] <= n and nums[i] != nums[nums[i]-1]:
            j = nums[i] - 1 # 减一是因为数组下标从0开始
            nums[i], nums[j] = nums[j], nums[i]
    
    # 找第一个学号与座位编号不匹配的学生
    for i in range(n):
        if nums[i] != i+1:
            return i+1
    
    # 所有学生都坐在正确的座位上
    return n+1
```

```python
nums = [7,8,9,11,12]
firstMissingPositive(nums)
```
