```python
from node import *
from heapq import *

def mergeKLists(lists):
    dummy = cur = ListNode()
    heap = []
    
    # 初始化堆，存储每个链表的头节点
    for i, node in enumerate(lists):
        if node:
            heappush(heap, (node.val, i))
    
    # 逐步取出最小节点
    while heap:
        val, i = heappop(heap)
        cur.next = lists[i]
        cur = cur.next
        if lists[i].next:
            lists[i] = lists[i].next
            heappush(heap, (lists[i].val, i))
    
    return dummy.next
```

```python
# 上述代码的升级版 改写内置方法实现节点的比较 直接将节点入堆

from node import *
from heapq import *
ListNode.__lt__ = lambda a,b: a.val < b.val # 让堆可以比较节点大小

def mergeKLists(lists):
    cur = dummy = ListNode()
    h = [head for head in lists if head] # 头节点入堆
    heapify(h)
    # 循环直到堆为空
    while h:
        node = heappop(h)
        if node.next:
            heappush(h, node.next)
        cur.next = node
        cur = cur.next
    return dummy.next
```

```python
# 先合并两个链表 再将第三个链表和新链表合并
def mergeTwoLists(list1, list2):
    cur = dummy = ListNode()
    while list1 and list2:
        if list1.val < list2.val:
            cur.next = list1
            list1 = list1.next
        else:
            cur.next = list2
            list2 = list2.next
        cur = cur.next
    cur.next = list1 or list2
    return dummy.next

# 递归写法 先合并左边 再合并右边
def mergeKLists(lists):
    m = len(lists)
    if m == 0:
        return None
    if m == 1:
        return lists[0]
    left = mergeKLists(lists[:m//2])
    right = mergeKLists(lists[m//2:])
    return mergeTwoLists(left, right)

# 迭代写法 先二二合并 再四四合并
def mergeKLists(lists) -> Optional[ListNode]:
    m = len(lists)
    if m == 0:
        return None
    step = 1
    while step < m:
        for i in range(0, m - step, step * 2):
            lists[i] = self.mergeTwoLists(lists[i], lists[i + step])
        step *= 2
    return lists[0]
```
