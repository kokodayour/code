```python
from node import *
from heapq import *

def mergeKLists(lists):
    dummy = cur = ListNode()
    heap = []
    
    # 初始化堆，存储每个链表的头节点
    for i, node in enumerate(lists):
        if node:
            heappush(heap, (node.val, i))
    
    # 逐步取出最小节点
    while heap:
        val, i = heappop(heap)
        cur.next = lists[i]
        cur = cur.next
        if lists[i].next:
            lists[i] = lists[i].next
            heappush(heap, (lists[i].val, i))
    
    return dummy.next
```

```python
# 上述代码的升级版 改写内置方法实现节点的比较 直接将节点入堆

from node import *
from heapq import *
ListNode.__lt__ = lambda a,b: a.val < b.val # 让堆可以比较节点大小

def mergeKLists(lists):
    cur = dummy = ListNode()
    h = [head for head in lists if head] # 头节点入堆
    heapify(h)
    # 循环直到堆为空
    while h:
        node = heappop(h)
        if node.next:
            heappush(h, node.next)
        cur.next = node
        cur = cur.next
    return dummy.next
```
