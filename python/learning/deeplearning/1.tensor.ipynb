##### 创建张量

```python
torch.tensor([2.0]) # 创建一个包含2.0的张量
torch.zeros((2,3)) # 创建两行三列的全零张量
torch.randn((2,3)) # 创建两行三列服从正态分布的随机张量
torch.arange(10) # 创建一个包含0到9的张量
```

#### 张量的变换

```python
a = torch.rand((2,3))
a.reshape(-1,1) # 改变张量的形状 -1表示自动计算维度
a.T # 转置向量
a.transpose(0,1) # 更换向量的第0维和第1维
a.view(6) # 返回视图 改变形状
```

##### 张量的数学运算

```python
torch.pow(torch.tensor([2.0, 3.0]), 3) # 计算2和3的3次方
torch.pow(3, torch.tensor([2.0, 3.0])) # 计算3的2和3次方
torch.pow(torch.tensor([2.0, 3.0]), torch.tensor([2.0, 3.0])) # 计算2的2次方和3的3次方
```
