```python
import torch
from torch import nn
```

```python
# nn.Parameter 模型的参数 参与反向传播
class MyLinear(nn.Module):
    def __init__(self, in_feature, out_feature):
        super().__init__()
        self.weight = nn.Parameter(torch.randn(in_feature, out_feature))
        self.bais = nn.Parameter(torch.randn(out_feature))
    def forward(self, src):
        return src@self.weight + self.bais
src = torch.randn((3,3))
net = MyLinear(3,4)
net(src)
net.bais # 模型偏置
net.weight # 模型权重
```

```python
# Dropout 输出形状等于输入
dropout = nn.Dropout(p=0.1)
a = torch.ones((2, 3))
dropout(a)
```

```python
# embedding 输入(句子个数, 句子长度) 输出(句子个数, 句子长度, 词向量维度)
embedding = nn.Embedding(num_embeddings=1000, embedding_dim=512) # 假设词典大小为1000，词向量维度为512
input_ids = torch.tensor([[1, 2, 4], [4, 3, 2]])
embedding(input_ids)
```

```python
# 层归一化 输入(x,y,z) 输出(x,y,z)
layernorm = nn.LayerNorm(normalized_shape=3, eps=0.00001)
x= torch.randn((2,2,3))
layernorm(x)
```

```python
# 多头注意力 输入(x, y, z) 输出(x, y, z)
heads_attention = nn.MultiheadAttention(embed_dim=512, num_heads=8, batch_first=True)
q = torch.randn((8, 64, 512))
k = torch.randn((8, 64, 512))
v = torch.randn((8, 64, 512))
outputs, weight = heads_attention(q,k,v)
```

```python
# Transformer Encoder 输出形状等于输入
encoder_layer = nn.TransformerEncoderLayer(
    d_model=16,
    nhead=8,
    dim_feedforward=2048,
    dropout=0.1,
)

# 堆叠多个Encoder
encoder = nn.TransformerEncoder(
    encoder_layer=encoder_layer,
    num_layers=6
    )
src = torch.randn((1, 4, 16))
encoder(src)
```

```python
# MSE损失层
inputs = torch.tensor([2.0, 3.0, 4.0])
outputs = torch.tensor([2.1, 2.8, 4.3])
# reduction指定损失聚合的方式 默认mean 也有sum none()
criterion = nn.MSELoss(reduction='mean')
loss = criterion(inputs, outputs)
loss.item()
```
