```python
def reverseKGroup(head, k):
    n = 0
    temp = head
    while temp:
        temp = temp.next
        n += 1

    p0 = dummy = ListNode(next = head)
    old = head
    while n >= k:
        new = ListNode(next = old)
        for _ in range(k):
            temp = old.next
            old.next = new
            new = old
            old = temp
        temp = p0.next
        p0.next.next = old
        p0.next = new
        p0 = temp
        n -= k
    return dummy.next
```

```python
def reverseKGroup(head, k: int):
    dummy = pre = ListNode(next=head)
    while pre.next:
        cur = pre.next
        for _ in range(k-1):
            nxt = cur.next
            cur.next = nxt.next
            nxt.next = pre.next
            pre.next = nxt
        temp = pre
        # determine whether there are remaining K elements
        for _ in range(k):
            temp = temp.next
            if not temp:
                return dummy.next  
    return dummy.next
```
