```python
def detectCycle(head):
    a = set()
    while head:
        if head.val in a:
            return head
        a.add(head.val)
        head = head.next
    return None
```

```python
# fast and slow pointer
def detecCycle(head):
    fast = slow = head
    while fast and fast.next:
        fast = fast.next.next
        slow = slow.next
        # there is a ring
        if fast is slow:
            while head is not slow:
                head = head.next
                slow = slow.next
            return slow
    return None
```
