```python
# find the middle node of linkedlist
def findMiddle(head):
    slow = fast = head
    while fast and fast.next:
        fast = fast.next.next
        slow = slow.next
    return slow

# reverse the Linkedlist
def reverseLinkedlist(head):
    new = None
    old = head
    while old:
        temp = old.next
        old.next = new
        new = old
        old = temp
    return new
```

```python
def reorderList(head):
    """
    Do not return anything. Modify head in-place instead
    """
    middle = findMiddle(head)
    second = reverseLinkedlist(middle)
    while second is not middle:
        temp = head.next
        head.next = second
        head = temp
        temp = second.next
        second.next = head
        second = temp
```
