```python
# tail insert
def reverseList(head):
    if (head is None) or (head.next is None):
        return head
    # the head of new linkedlist
    new = reverseList(head.next)
    # the tail of new linkedlist
    tail = head.next
    # reverse the direction between tail and head
    tail.next = head
    head.next = None
    return new
```

```python
# head insert
def reverseList(head):
    old = head
    new = None
    while old:
        temp = old.next
        old.next = new
        new = old
        old = temp
    return new
```
