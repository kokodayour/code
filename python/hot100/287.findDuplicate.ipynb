```python
# cannot modify nums and use O(1) space
# refer to 142
def findDuplicate(nums):
    # 0 must not be in the ring
    slow = fast = 0
    fast_1 = []
    slow_1 = []
    while True:
        slow = nums[slow] # equal to slow = slow.next
        fast = nums[nums[fast]] # equal to fast = fast.next.next
        fast_1.append(fast)
        slow_1.append(slow)
        if fast == slow: # fast and slow pointer move to the same node
            break

    head = 0
    print(slow_1)
    print(fast_1)
    while slow != head:
        slow = nums[slow]
        head = nums[head]
    return slow
```

```python
nums = [1,3,4,2,2]
findDuplicate(nums)
```
